```python
import os
os.environ["IPFS_GATEWAY"] = "https://ipfs.orcestra-campaign.org"

import xarray as xr

cid = "bafybeiesyutuduzqwvu4ydn7ktihjljicywxeth6wtgd5zi4ynxzqngx4m"

ds = xr.open_dataset(
    f"ipfs://{cid}",
    engine="zarr",
)

ds
```

```python
ds.launch_time.values
```

```python
ds.launch_time

launch_time_slices = [slice(t - np.timedelta64(5, 's'), t + np.timedelta64(5, 's')) for t in ds.launch_time.values]
```

```python
ds_velox_dropsonde = []


for launch_time_slice in tqdm(launch_time_slices):
    try: 
        ds_velox = campaign.pushbroom(launch_time_slice, georef=True)
        ds_velox_dropsonde.append(ds_velox)
    except Exception as e:
        print(f"Failed to load velox for launch time slice {launch_time_slice}: {e}")
```
